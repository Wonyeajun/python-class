```python
# 온실가스 지도(연도 슬라이더 + 기체 토글 + 등급 색상)

import pandas as pd
import numpy as np
import plotly.graph_objects as go
import pycountry
# → 필요한 라이브러리 불러오기


# -----------------------------------------------------
# 설정: DataFrame 변수명을 확인하세요
# -----------------------------------------------------
# All_GHG, CO2, CH4, N2O 라는 이름의 데이터프레임이 이미 존재한다고 가정
# index 또는 Country 컬럼에 국가명이 있어야 함
# CSV에서 불러올 경우 예: All_GHG = pd.read_csv(...)
# -----------------------------------------------------


# -----------------------------------------------------
# 예비: 만약 Country가 인덱스에 있으면 컬럼으로 내리기
# -----------------------------------------------------
def ensure_country_col(df):
    df = df.copy()                      # 원본 훼손 방지 위해 복사
    # 인덱스 이름이 'Country'이거나 Country 컬럼이 없고, 인덱스가 문자열(object)이면
    if df.index.name == 'Country' or ('Country' not in df.columns and df.index.dtype == object):
        df = df.reset_index()           # 인덱스를 일반 컬럼으로 되돌림
    if 'Country' not in df.columns:
        # 그래도 Country 컬럼이 없다면 문제 → 오류 발생
        raise ValueError("DataFrame에 'Country' 컬럼이 없습니다.")
    return df

All_GHG = pd.read_csv('historical_emissions.csv')
# 각 데이터프레임에 Country 컬럼이 존재하는지 확인하고 보정
All_GHG = ensure_country_col(All_GHG)
# CO2     = ensure_country_col(CO2)
# CH4     = ensure_country_col(CH4)
# N2O     = ensure_country_col(N2O)


# -----------------------------------------------------
# 1) 국가명 → ISO3 변환 함수
# -----------------------------------------------------
def country_to_iso3(name):
    if pd.isna(name):         # 값이 NaN이면 None 반환
        return None
    try:
        return pycountry.countries.lookup(name).alpha_3  # 정확한 이름 검색
    except Exception:
        try:
            # fuzzy search (유사 이름 검색)
            return pycountry.countries.search_fuzzy(name)[0].alpha_3
        except Exception:
            return None       # 실패시 None


# 각 DF에 iso3 컬럼 추가
# for df in [All_GHG, CO2, CH4, N2O]:
for df in [All_GHG]:
    df['iso3'] = df['Country'].apply(country_to_iso3)


# 공통적으로 iso3가 존재하는 국가들의 교집합을 계산
# common_iso3 = set(All_GHG['iso3'].dropna()) & set(CO2['iso3'].dropna()) & \
#               set(CH4['iso3'].dropna()) & set(N2O['iso3'].dropna())
common_iso3 = set(All_GHG['iso3'].dropna())
print(f"공통 사용 국가 수: {len(common_iso3)}")

# 공통 iso3를 가진 행만 사용하도록 필터링
All_GHG = All_GHG[All_GHG['iso3'].isin(common_iso3)].copy().reset_index(drop=True)
# CO2     = CO2[CO2['iso3'].isin(common_iso3)].copy().reset_index(drop=True)
# CH4     = CH4[CH4['iso3'].isin(common_iso3)].copy().reset_index(drop=True)
# N2O     = N2O[N2O['iso3'].isin(common_iso3)].copy().reset_index(drop=True)
All_GHG
```
